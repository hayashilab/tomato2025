```python
# Train the model
results = model.train(data="data.yaml",epochs=100,imgsz=640 ,optimizer='auto',patience=100,lr0=0.01,dropout=0.0, weight_decay=0.0005 ,device=0, plots=True)
```

```python

# Load a model
model = YOLO("runs/segment/train/weights/best.pt")  # load a custom model

# Validate the model
metrics = model.val()  # no arguments needed, dataset and settings remembered
metrics.box.map  # map50-95(B)
metrics.box.map50  # map50(B)
metrics.box.map75  # map75(B)
metrics.box.maps  # a list contains map50-95(B) of each category
metrics.seg.map  # map50-95(M)
metrics.seg.map50  # map50(M)
metrics.seg.map75  # map75(M)
metrics.seg.maps  # a list contains map50-95(M) of each category
```

```python
from PIL import Image
# Load a model
model = YOLO("runs/segment/train/weights/best.pt")  # load a custom model

# Predict with the model
results = model("picture/5.jpg")  # predict on an image

# Visualize the results
for i, r in enumerate(results):
    # Plot results image
    im_bgr = r.plot()  # BGR-order numpy array
    im_rgb = Image.fromarray(im_bgr[..., ::-1])  # RGB-order PIL image

    # Show results to screen (in supported environments)
    r.show()

    # Save results to disk
    r.save(filename=f"results{i}.jpg")
```
